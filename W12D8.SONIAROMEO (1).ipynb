# `new_cases`

```python
df_covid["date"] = pd.to_datetime(df_covid["date"])
```

```python
df_continents = df_covid[df_covid["continent"].notna()].copy()
```

```python
# Estraggo l'anno dalla data
df_continents["year"] = df_continents["date"].dt.year
```

```python
#Sommo per anno e per contintente
cases_year_continent = (
    df_continents.groupby(["year", "continent"])["new_cases"]
    .sum()
    .reset_index()
)
cases_year_continent
```

```python
#Calcolo solo per anno
world_cases_year = (
    df_continents.groupby("year")["new_cases"]
    .sum()
    .reset_index()
    .rename(columns={"new_cases": "world_cases"})
)
world_cases_year
```

```python
#Creo una tabella pivot
tot_casi_per_anno = pd.pivot_table(
    df_continents,
    values="new_cases",
    index="continent",
    columns="year",
    aggfunc="sum"
)
#Aggiungo la colonna 'Totale' per ogni continente e ka colonna 'Totale mondiale' per ogni anno
tot_casi_per_anno["Totale"] = tot_casi_per_anno.sum(axis=1)
tot_casi_per_anno.loc["Totale mondiale"] = tot_casi_per_anno.sum(axis=0)

print ("Totale dei contagi Covid-19 per anno")
tot_casi_per_anno 
```

```python
#Totale mondiale per anno (prendo la riga Totale mondiale e tolgo la colonna Totale)
world_year = tot_casi_per_anno.loc["Totale mondiale"].drop("Totale")

#Percentuali: (casi continente anno/totale mondo anno)* 100
pct_tot_casi_per_anno = tot_casi_per_anno.drop(index="Totale mondiale").drop(columns="Totale").div(world_year, axis=1) * 100

world_total = tot_casi_per_anno.loc["Totale mondiale", "Totale"] 
pct_tot_casi_per_anno["Media_%"] = (tot_casi_per_anno["Totale"] / world_total) * 100

#Riga Totale mondiale = 100% per ogni anno (e 100 anche per Media_%)
pct_tot_casi_per_anno.loc["Totale mondiale"] = 100  
pct_tot_casi_per_anno = pct_tot_casi_per_anno.round(2)
print ("Totale percentuale dei contagi Covid-19 per anno")
pct_tot_casi_per_anno
```

## Interpretazione delle variabili vaccinali

L’analisi delle colonne `new_vaccinations` e `total_vaccinations`,
osservata su base temporale, evidenzia che:

- `new_vaccinations` rappresenta un flusso giornaliero;
- `total_vaccinations` rappresenta un cumulato crescente.

Il totale delle vaccinazioni può essere derivato sia aggregando i flussi giornalieri sia considerando il valore massimo del cumulato.

```python
#Aggrego le vaccinazioni giornaliere (new_vaccinations) per continente e anno mediante pivot_table

pivot_vacc = pd.pivot_table(
    df_continents,
    values="new_vaccinations",
    index="continent",
    columns="year",
    aggfunc="sum"
)

#Calcolo il totale delle vaccinazioni per ciascun continente
pivot_vacc["Totale"] = pivot_vacc.sum(axis=1)

#Calcolo il totale mondiale per ciascun anno
pivot_vacc.loc["Totale mondiale"] = pivot_vacc.sum(axis=0)

print ("Totale delle vaccinazioni Covid-19 per anno")
pivot_vacc
```

```python
world_vacc_year = pivot_vacc.loc["Totale mondiale"].drop("Totale")

#Estraggo il totale mondiale per ciascun anno eliminando la colonna 'Totale' perché mi servono solo i totali annuali

pivot_vacc_pct = (
    pivot_vacc.drop(index="Totale mondiale") 
              .drop(columns="Totale") 
              .div(world_vacc_year, axis=1) * 100
)

pivot_vacc_pct["% media"] = pivot_vacc_pct.mean(axis=1).round(2) 
pivot_vacc_pct.loc["Totale mondiale"] = (100) 

pivot_vacc_pct=pivot_vacc_pct.round(2)
print ("Totale percentuale delle vaccinazioni Covid-19 per anno")
pivot_vacc_pct
```

## Per approfondire il confronto tra i Paesi e mantenere coerenza nell’analisi delle dinamiche ospedaliere, può essere considerata la variabile alternativa `weekly_hosp_admissions`, che rappresenta i nuovi ingressi ospedalieri su base settimanale ed è valorizzata per tutte le nazioni considerate.

weekly_hosp_admissions:	Number of COVID-19 patients newly admitted to hospitals in a given week (reporting date and the preceeding 6 days)

```python
#Seleziono le colonne legate all'ospedalizzazione/ricoveri per verificare la disponibilità dei dati
cols_hosp = [
    "hosp_patients",
    "hosp_patients_per_million",
    "weekly_hosp_admissions",
    "weekly_hosp_admissions_per_million",
    "weekly_icu_admissions"
]

df_germany_2021 = df_covid[
    (df_covid["location"] == "Germany") &
    (df_covid["date"].dt.year == 2021)
]

df_germany_2021[cols_hosp].isna().sum() #Conteggio dei valori mancanti (NaN) per ciascuna colonna selezionata
```

df_germany_2021[cols_hosp].notna().sum() #Conteggio dei valori presenti (non-NaN) per ciascuna colonna selezionata

```python
#Verifico la distribuzione dei dati di "cols_hops".
df_germany_2021[cols_hosp].describe()
```

```python
df_germany_2021[["date", "weekly_hosp_admissions"]].head(20)
```

```python
#Converto la colonna `date` in formato datetime per poter estrarre correttamente anno e settimana
df_hosp_2021["date"] = pd.to_datetime(df_hosp_2021["date"])
df_hosp_2021["year"] = df_hosp_2021["date"].dt.year
df_hosp_2021["week"] = df_hosp_2021["date"].dt.isocalendar().week
```

```python
#Ricostruisco una serie settimanale selezionando una sola osservazione rappresentativa per settimana e per Paese
weekly_real = (
    df_hosp_2021
    .sort_values("date")
    .groupby(["location", "year", "week"])["weekly_hosp_admissions"]
    .last() #Dopo l’ordinamento per data, ".last()" prende l’ultima rilevazione disponibile della settimana
)
```

```python
#Verifico il numero di settimane disponibili per ciascun Paese
weekly_real.groupby("location").size()
```

```python
#Calcolo il totale annuo dei ricoveri per ciascun Paese come somma dei valori settimanali
tot_2021 = weekly_real.groupby("location").sum()
tot_2021
```

```python
plt.figure()
tot_2021.plot(kind="bar", color=["blue", "red", "orange", "green"])
plt.title("Ricoveri COVID-19 – Totale 2021 (dati settimanali)")
plt.xlabel("Paese")
plt.ylabel("Numero totale di ricoveri")
plt.tight_layout()
plt.show()
```

```python
plt.figure()

ax = weekly_real.unstack("location").plot(color=["blue", "red", "orange", "green"])
ax.set_title("Trend settimanale dei ricoveri – 2021")
ax.set_xlabel("Anno, settimana")
ax.set_ylabel("Numero di ricoveri settimanali")
ax.legend(title="Paese")
plt.tight_layout()
plt.show()
```
